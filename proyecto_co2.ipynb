```python
import pandas as pd
url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
df = pd.read_csv(url)
df_limpio = df[df['year'] >= 1950].dropna(subset=['co2'])
print("¡Datos cargados con éxito! Total de registros:", len(df_limpio))
```

```python
import matplotlib.pyplot as plt

# Filtrar países de interés usando el DataFrame limpio
paises = ['Colombia', 'United States', 'China', 'Brazil']
df_paises = df_limpio[df_limpio['country'].isin(paises)]

# Configurar la gráfica
plt.figure(figsize=(10, 6))

for pais in paises:
    datos_pais = df_paises[df_paises['country'] == pais]
    plt.plot(datos_pais['year'], datos_pais['co2'], label=pais, linewidth=2)

plt.title('Evolución de Emisiones de CO2 (1950 - Actualidad)', fontsize=14)
plt.xlabel('Año', fontsize=12)
plt.ylabel('Emisiones de CO2 (millones de toneladas)', fontsize=12)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

## Conclusiones del Análisis de Emisiones de CO2

1. **Crecimiento histórico acelerado:** A partir de 1950 se observa un incremento sostenido en las emisiones globales de CO2 impulsado por la industrialización y el crecimiento demográfico.
2. **Diferencias entre economías:** Países como China y Estados Unidos concentran los mayores volúmenes de emisiones históricas y recientes, mientras que países latinoamericanos como Colombia y Brasil mantienen niveles considerablemente menores en comparación global.
3. **Importancia de la limpieza de datos:** Filtrar registros anteriores a 1950 y excluir agregados continentales permitió analizar el comportamiento real a nivel país sin distorsiones estadísticas.
