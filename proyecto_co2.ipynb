```python
 import pandas as pd
import numpy as np

# Cargar el dataset directamente desde la fuente oficial de Our World in Data
url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
df = pd.read_csv(url)

# Visualizar las primeras filas del dataframe
df.head()
```

```python
  # Definir variables de interés
columnas = ['country', 'year', 'co2', 'co2_per_capita', 'gdp']
df_filtrado = df[columnas].copy()

# Filtrar registros a partir del año 1950 y excluir agregados globales/continentales
df_paises = df_filtrado[(df_filtrado['year'] >= 1950) &
                        (~df_filtrado['country'].isin(['World', 'High-income countries', 'Asia', 'Europe', 'North America', 'Africa']))]

# Eliminar filas con datos nulos en la variable principal (co2)
df_limpio = df_paises.dropna(subset=['co2']).reset_index(drop=True)

# Verificar resumen
print(f"Total de registros limpios: {len(df_limpio)}")
df_limpio.info()
```
